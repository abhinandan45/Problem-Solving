```python
def fact(num):

    if num == 1:
        return 1
    
    else:
        return num * fact(num-1)
    

num = 5
print(fact(num))
    
```

```python
def fib(numb):
    if numb == 0:
        return 0
    elif numb == 1:
        return 1
    else:
        return fib(numb-1) + fib(numb-2)

numb = 10

print(fib(numb))
```

```python
la = [1, 2, 3, 4, 5]

def rev_lis(la):
    if len(la) == 0:
        return la          
    else:
        return rev_lis(la[1:]) + [la[0]]    

print(rev_lis(la))
```
