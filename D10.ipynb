```python
def has_duplicate(l):
    s = set()
    for i in l:
        if i in s:
            return True

        else:
            s.add(i)
    return False

l = [1,2,3,4,5,6,7,7]
print(has_duplicate(l))
```

```python
def two_sum(nums,target):
    seen = {}

    for i,x in enumerate(nums):
        need = target - x
        if need in seen:
            return seen[need] , i

        seen[x] = i

    

nums = [2,7,11,15]
target = 9

print(two_sum(nums,target))    
```

```python
def is_ana(p,q):
    c = {}
    d = {}

    for i in p:
        if i not in c:
            c[i] =1

        else:
            c[i] += 1

    for j in q:
        if j not in d:
            d[j] = 1

        else:
            d[j] += 1


    return True if c==d else False



p = "ab"
q = "abc"

print(is_ana(p,q))
```
