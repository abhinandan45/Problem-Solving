```python
num = 5

fac = 1

for i in range(1,num+1):
    fac *= i

print(fac)
```

```python
def factorial(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * factorial(num-1)
    
print(factorial(5))
```

```python
num = 11

prime = True

if num < 2:
    prime = False

else:
    for i in range(2,num):
        if num % i == 0:
            prime = False
            break

if prime:
    print(num, "is a prime")

else:
    print(num, "is not a prime")
```
