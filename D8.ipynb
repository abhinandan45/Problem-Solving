```python
class student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

    def is_pass(self):
        if self.marks >= 33:
            return True

        else:
            return False


s1 = student("abhi",45)

s1.is_pass()
```

```python
def powerr(a, b):
    if b == 0:
        return 1
    else:
        return a * powerr(a, b-1)

a, b = 2, 5
print(powerr(a, b))
```
