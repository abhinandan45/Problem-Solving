```python
def sec_large(l):
    large = float('-inf')
    second_large = float('-inf')

    for i in l:
        if i > large:
            second_large = large
            large = i

        elif i > second_large and i < large:
            second_large = i


    return second_large

l = [5,5,3]
print(sec_large(l))
```

```python
def palindrome(s):
    a = s.lower().replace(" ", "")

    rev = ""

    for i in a:
        rev = i + rev

    return "palin" if a == rev else "not palin"

print(palindrome("hello Olleh"))
```

```python
def find_pairs(lis,target):
    d = []


    for i in lis:
        for j in lis:
            if i + j == target:
                d.append((i,j))
    return d


lis = [1,2,3,4,5,6]
target = 5

print(find_pairs(lis,target))

```
