```python

sen = "Madam"
ak = sen.lower()
new = ""

for i in ak:
    if i != " ":
        new += i

rev = ""

for j in new:
    rev = j + rev

print(rev)
print(new)
if new == rev:
    print(sen, "is a palindrome")

else:
    print(sen, "sen is not a palindrome")
```

```python
def sum_list(lis):
    if len(lis) == 0:
        return 0
    else:
        return lis[0] + sum_list(lis[1:])


print(sum_list([1,2,3]))

        

    
```

```python
a = [1,2,3,4]
b = [3,4,5,6]

common = []
for i in a:
    if i in b:
        common.append(i)

print(common)
```

```python
arr = [12, 35, 1, 10, 34, 1]

large = 0
sec_large = 0

for i in arr:
    if i > large:
        sec_large = large
        large = i
    elif i > sec_large:
        sec_large = i

print(sec_large)
```

```python
large = float('-inf')
sec_large = float('-inf')

for i in arr:
    if i > large:
        sec_large = large
        large = i
    elif i > sec_large and i != large:
        sec_large = i

print(sec_large)
```
