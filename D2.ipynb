```python
b = 12
a = []
c = 18
d = []

for i in range(1,b+1):
    if b % i == 0:
        a.append(i)

for j in range(1,c+1):
    if c % j == 0:
        d.append(j)

print(f"{b} divisers = {a} , {c} divisers = {d}")

large = 0

for k in d:
    if k > large and k in a:
        large = k

print(large)
```
