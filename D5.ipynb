```python
l = [1, 2, 3, 4, 3,3,3, 2]
freq = {}

for i in l:
    if i in freq:
        freq[i] += 1

    else:
        freq[i] = 1

j = 0
m = 0
for k,l in freq.items():
    if l > j:
        j = l
        m = k


print(f"{m} -> {j}")
```

```python
num = 11
prime = True

if num < 2:
    prime = False

else:
    prime = True
    for i in range(2,num):
        if num % i == 0:
            prime = False
            break


if prime:
    print(num,"is prime")

else:
    print(num, "is not a prime")
```
