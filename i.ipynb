```python
def revers_str(a):
    k = ""

    for i in a.lower().replace(" ",""):
        
        k = i + k

    return k

a = "abhinandan"
print(revers_str(a))
```

```python
b = "rama"

def chech_pali(b):
    d = ""

    for i in b:
        d = i + d

    return d == b

re = (chech_pali(b))

if re:
    print(b,"is palindrome")
else:
    print(b,"is not a palindrome")
```

```python
sen = "abhinandan"

freq = {}

for i in sen:
    if i not in freq:
        freq[i] = 1

    else:
        freq[i] += 1

print(freq)
```

```python
nums = [1,2,3,4,2,5,6,3,4]

a = []

duplicates = []

for i in nums:
    if i not in a:
        a.append(i)

    else:
        duplicates.append(i)
    
print(duplicates)
```

```python
num = [1,2,5,6,8,9,10,10,4,9]

large = float('-inf')
sec_large = float('-inf')

for i in num:
    if i > large:
        sec_large = large
        large = i

    if i > sec_large and i != large:
        sec_large = i

print(sec_large)
```

```python
n = 20

for i in range(0,n+1):
    if i % 3 == 0:
        print("fizz")

    if i % 5 == 0:
        print("buzz")

    if i % 3 ==0 and i % 5 == 0:
        print("FizzBuzz")

    else:
        print(i)
```

```python
sent = "my name is abhinandan"

vowels = "aeiou"

count = 0

for i in sent.lower():
    if i in vowels:
        count += 1

print(count)
```

```python
n = 10

is_p = True

if n <= 1:
    is_p = False

else:
    for i in range(2,n):
        if n % i == 0:
            is_p = False


if is_p:
    print(n,"is prime")

else:
    print(n,"is not a prime")
```
