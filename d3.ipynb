```python
a = [1, 2, 2, 3, 4, 4, 5]

uniques = []

for i in a:
    if i not in uniques:
        uniques.append(i)

print(uniques)
```

```python
a = "listen"
b = "silent"
anagram = True

if len(a) != len(b):
    anagram = False
else:
    for i in a:
        if a.count(i) != b.count(i):
            anagram = False


if anagram:
    print(f"{a} and {b} are anagrams")
else:
    print(f"{a} and {b} are not anagrams")


```

```python
l = [5, 2, 8, 1, 9]
n = len(l)

for i in range(n):
    for j in range(0,n-1):
        if l[j] > l[j+1]:
           l[j], l[j+1] = l[j+1], l[j] 

print(l)
```

```python
l = [4, 6, 1, 9, 3]

p = len(l)

for i in range(p):
    for k in range(0,p-1):
        if l[k] > l[k+1]:
            l[k], l[k+1] = l[k+1], l[k]
        print(f"{l[k]} {l[k+1]}")

print(l)
```

```python
a = [1, 3, 5]
b = [2, 4, 6]

c = a+b

d = len(c)

for i in range(d):
    for l in range(0,d-1):
        if c[l] > c[l+1]:
            c[l] , c[l+1] = c[l+1], c[l]

print(c)
```

```python
a = [1, 3, 5]
b = [2, 4, 6]
c = []

i = 0   
j = 0   

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        c.append(a[i])
        i += 1
    else:
        c.append(b[j])
        j += 1

while i < len(a):
    c.append(a[i])
    i += 1

while j < len(b):
    c.append(b[j])
    j += 1

print(c)
```
