```python
ab = "the quick brown fox jumped"

k = ab.split()

word = ""

long = 0

for i in k:
    if long < len(i):
        long = len(i)
        word = i

print(word)
print(long)
```

```python
class rectangle:
    def __init__(self,length,width):

        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2*(self.length+self.width)

ak = rectangle(10,20)
print(ak.area())
```
