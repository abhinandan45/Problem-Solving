```python
n = 123

sum = 0

while n > 0:
    i = n % 10
    sum += i
    n //= 10

print(sum)
```

```python
def digit_sum(num):

    if num == 0:
        return 0
    else:
        return num % 10 + digit_sum(num//10)

print(digit_sum(12))
```

```python
matrix = [[1,2,3],[4,5,6]]
print(matrix[0][1])

rows = len(matrix)        # 2
cols = len(matrix[0])     # 3

result = []
for i in range(cols):          # naya matrix me utni rows honi chahiye jitne purane me columns the
    new_row = []
    for j in range(rows):
        new_row.append(matrix[j][i])   # yahan dhyan se dekh index kaise swap hue
    result.append(new_row)

print(result)
```

```python
mat = [[1,2,3],[4,5,6]]

row = len(mat)
col = len(mat[0])
print(row)
print(col)

re = []

for i in range(col):
    new = []
    for j in range(row):
        new.append(mat[j][i])
    re.append(new)

print(re)
```
