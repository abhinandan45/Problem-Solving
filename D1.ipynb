```python
num = int(input("enter a num : "))
nb = num
rev = 0

while num > 0:
    n = num % 10
    rev = (rev*10) + n
    num //= 10

if nb == rev:
    print(f"this {rev} is palindrome ")

else:
    print(f"this {rev} is not a palindrome ")

```

```python
l = [10, 5, 8, 20, 15]
large = l[0]
second_large = 0

for i in l:
    if i > large:
        second_large = large
        large = i

    elif i > second_large and i != large:
        second_large = i

print(second_large)

```
