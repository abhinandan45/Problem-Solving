```python
def group_anagrams(strs):
    d = {}
    for i in strs:
        key = ''.join(sorted(i))    # sorted key banao
        if key not in d:
            d[key] = [i]             # nayi key, list mein pehla item daal do
        else:
            d[key].append(i)         # existing key, list mein add kar do
    return list(d.values())   
    
strs = ["eat","tea","tan","ate","nat","bat"]
print(group_anagrams(strs))
```

```python
strs = ["eat","tea","tan","ate","nat","bat"]

d = {}

for i in strs:
    k = "".join(sorted(i))

    if k not in d:
        d[k] = [i]

    else:
        d[k].append(i)

print(list(d.values()))
```
